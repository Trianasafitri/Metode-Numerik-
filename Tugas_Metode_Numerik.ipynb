## Persamaan yang Digunakan

f(t) = 800e^(0.3t) - 2000 - 400t = 0

Metode: Regula Falsi
Interval awal: a = 5 dan b = 6
Toleransi: 0.0001


```python

import math

def f(t):
    return 800 * math.exp(0.3*t) - 2000 - 400*t

a, b = 5.0, 6.0
tol = 0.0001
prev = None

print("ITERASI REGULA FALSI")
print(f"{'Iterasi':<8}{'a':<12}{'b':<12}{'c':<12}{'f(c)':<12}{'Error':<12}")

for i in range(1, 101):
    c = (a*f(b) - b*f(a)) / (f(b) - f(a))
    fc = f(c)
    error = abs(c-prev) if prev is not None else 0

    print(f"{i:<8}{a:<12.4f}{b:<12.4f}{c:<12.4f}{fc:<12.4f}{error:<12.4f}")

    if abs(fc) < tol:
        break

    if f(a)*fc < 0:
        b = c
    else:
        a = c

    prev = c

print(f"\nTitik impas = {c:.4f} bulan")
print(f"Jumlah iterasi = {i}")
```
